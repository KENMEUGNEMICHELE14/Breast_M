```python
import pandas as pd

df=pd.read_csv('../data/data.csv')
df['diagnosis'] = df['diagnosis'].map({'M':1, 'B':0})# 1 pour maligne et 0 pour benigne
df
print(df['diagnosis'].value_counts()) # ici, value_counts() est utilisé pour compter le nombre d'occurrences de chaque valeur unique dans la colonne 'diagnosis'. Cela permet de voir combien de cas sont malins (1) et combien sont bénins (0).
```

```python
print("Les dimensions du DataFrame sont :" + str(df.shape)) # Ici, shape me donne 569 lignes et 33 colonnes
```

```python
# Les types de données des colonnes du DataFrame y compris le nombre de valeurs non nulles de chaque colonne
df.info() 
```

```python
# Ici, describe() est utilisé pour générer des statistiques descriptives du DataFrame. Cela inclut des mesures telles que la moyenne, l'écart type, les valeurs minimales et maximales, ainsi que les quartiles pour chaque colonne numérique du DataFrame.
if 'id' in df.columns:
    df =df.drop(columns=['id'])  # Supprimer la colonne 'id' si elle existe
df.describe()
```

```python
df.skew() # Ici, skew() est utilisé pour calculer le coefficient d'asymétrie de chaque colonne numérique du DataFrame. L'asymétrie mesure la symétrie de la distribution des données. Une valeur proche de 0 indique une distribution symétrique, tandis qu'une valeur positive ou négative indique une distribution asymétrique.et si elle est asymétriquen on peut appliquer une transformation pour la rendre plus symétrique. Cela peut être utile pour certaines analyses mathematiques pour equilibrer les données et améliorer la performance des modèles d'apprentissage automatique.
```
