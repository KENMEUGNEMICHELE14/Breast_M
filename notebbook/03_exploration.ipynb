```python
# Sélection des variables numériques (hors diagnosis)
cols_physiques = [c for c in df.columns if c != 'diagnosis']

outliers_summary = {}

for col in cols_physiques:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    
    # Bornes de détection
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    # Compter les points hors bornes
    nb_outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)].shape[0]
    outliers_summary[col] = nb_outliers

# Afficher le bilan pour toutes les colonnes
df_outliers = pd.DataFrame(list(outliers_summary.items()), columns=['Variable', 'Nombre d_outliers'])
print(df_outliers)
```

```python
# Extraction des paires de variables fortement corrélées entre elles (ex: > 0.85)
threshold = 0.85
high_corr_pairs = []

for i in range(len(corr_matrix.columns)):
    for j in range(i):
        # On exclut la variable cible 'diagnosis'
        col1 = corr_matrix.columns[i]
        col2 = corr_matrix.columns[j]
        if col1 != 'diagnosis' and col2 != 'diagnosis':
            val = corr_matrix.iloc[i, j]
            if abs(val) >= threshold:
                high_corr_pairs.append((col1, col2, round(val, 2)))

df_high_corr = pd.DataFrame(high_corr_pairs, columns=['Variable 1', 'Variable 2', 'Corrélation (r)'])
print(f"=== PAIRES DE VARIABLES REDONDANTES (|r| >= {threshold}) ===")
print(df_high_corr.to_string())
```

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
```

```python
# # 1. Définition des variables explicatives X (les 30 colonnes) et de la cible y
X = df.drop(columns=['diagnosis'])
y = df['diagnosis']

# 2. ÉTAPE 1 : Découpage Train (80 %) / Test (20 %) sur les 30 variables
X_train_all,X_test_all,Y_train,Y_test=train_test_split(
    X, # X est la valeur d'entrée pour les autres 30 variables
    y, # y est la valeur d'entrée(pour la colonne diagostics)
    test_size=0.20, # 20% des donnes seont dans les test et 80% dans le train
    stratify=y,  # Pourque la proportion des malignes et benine reste la meme utant dans le train que dans le test
    random_state=42 # Pour que à chaque fois que je run le jeu de  de train reste le meme , genre ne se change pas pour utiliser à un tour donné le jeu de test
)
```

```python
# Identifions les familles d evariables et trouvons le chef de famille qui a le plus de lisaion avec la variable diagnosis dans ce groupe de famille

# 3. Regroupement par familles (Pure Python)
familles = []
for col1, col2,_ in high_corr_pairs:
    trouve = False
    for f in familles:
        if col1 in f or col2 in f:
            f.add(col1)
            f.add(col2)
            trouve = True
            break
    if not trouve:
        familles.append({col1, col2})

# Fusion des sous-groupes interconnectés
familles_fusions = []
while familles:
    first, *rest = familles
    first = set(first)
    lf = -1
    while len(first) > lf:
        lf = len(first)
        rest2 = []
        for r in rest:
            if first & set(r):
                first |= set(r)
            else:
                rest2.append(r)
        rest = rest2
    familles_fusions.append(list(first))
    familles = rest

# 4. Élection du chef de famille
corr_with_target = df.corr()['diagnosis'].abs()

variables_a_conserver = []
variables_a_supprimer = set()

for famille in familles_fusions:
    chef = max(famille, key=lambda var: corr_with_target[var])
    variables_a_conserver.append(chef)
    for var in famille:
        if var != chef:
            variables_a_supprimer.add(var)

print("=== DÉCISION DE SÉLECTION ===")
print("Chefs de famille conservés :", variables_a_conserver)
print("Variables redondantes à supprimer :", list(variables_a_supprimer))
```

```python
#  Suppression des 6 variables redondantes APRÈS le split
# (Pour créer les jeux de données réduits X_train_red et X_test_red)
# 1. On récupère la liste exacte des 14 variables redondantes identifiées par l'algorithme
vars_a_supprimer = list(variables_a_supprimer)

# 2. Suppression des 14 variables redondantes APRÈS le train_test_split
X_train_red = X_train_all.drop(columns=vars_a_supprimer)
X_test_red = X_test_all.drop(columns=vars_a_supprimer)

# 3. Vérification des nouvelles dimensions
print("=== JEU COMPLET (Baseline) ===")
print(f"X_train_all : {X_train_all.shape}  |  X_test_all : {X_test_all.shape}")

print("\n=== JEU RÉDUIT (Sans multicolinéarité) ===")
print(f"X_train_red : {X_train_red.shape}  |  X_test_red : {X_test_red.shape}")
```

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler

# 1. Initialisation des scalers (un pour chaque expérience)
scaler_all = StandardScaler()
scaler_red = StandardScaler()

# -------------------------------------------------------------
# EXPÉRIENCE 1 : JEU COMPLET (30 variables)
# -------------------------------------------------------------
# A. Fit + Transform sur le Train
X_train_all_scaled = scaler_all.fit_transform(X_train_all)

# B. Transform SEULEMENT sur le Test
X_test_all_scaled = scaler_all.transform(X_test_all)

# Restitution sous forme de DataFrames pandas (facultatif mais plus pratique)
X_train_all_scaled = pd.DataFrame(X_train_all_scaled, columns=X_train_all.columns, index=X_train_all.index)
X_test_all_scaled = pd.DataFrame(X_test_all_scaled, columns=X_test_all.columns, index=X_test_all.index)

# -------------------------------------------------------------
# EXPÉRIENCE 2 : JEU RÉDUIT (16 variables)
# -------------------------------------------------------------
# A. Fit + Transform sur le Train
X_train_red_scaled = scaler_red.fit_transform(X_train_red)

# B. Transform SEULEMENT sur le Test
X_test_red_scaled = scaler_red.transform(X_test_red)

# Restitution sous forme de DataFrames pandas
X_train_red_scaled = pd.DataFrame(X_train_red_scaled, columns=X_train_red.columns, index=X_train_red.index)
X_test_red_scaled = pd.DataFrame(X_test_red_scaled, columns=X_test_red.columns, index=X_test_red.index)

# 2. Vérification rapide (Moyenne ≈ 0, Écart-type ≈ 1 sur le train)
print("=== VÉRIFICATION D'ÉCHELLE (X_train_all_scaled) ===")
print("Moyenne des variables (doit être ~0) :", round(X_train_all_scaled.mean().mean(), 4))
print("Écart-type des variables (doit être ~1) :", round(X_train_all_scaled.std().mean(), 4))
```
